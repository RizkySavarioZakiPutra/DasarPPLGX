```python

# Kasus: Ibu Guru punya 25 buah permen, ingin dibagikan rata ke 4 siswa.
jumlah_permen = 25
jumlah_siswa = 4

permen_per_siswa = jumlah_permen // jumlah_siswa
sisa_permen = jumlah_permen % jumlah_siswa

print(f"Setiap siswa mendapat : {permen_per_siswa} permen")
print(f"Sisa permen di toples : {sisa_permen} permen")

print("15 % 2 = ", 15 % 2)
print("14 % 2 = ", 14 % 2)
```

```python

print("=== PROGRAM KONVERSI SUHU CELCIUS KE FAHRENHEIT ===")

# 1. Ambil input suhu Celcius menggunakan float() agar bisa menerima angka desimal
celcius = float(input("Masukkan suhu dalam derajat Celcius: "))

# 2. Masukkan rumus: (celcius dikali 1.8) ditambah 32
fahrenheit = (celcius * 1.8) + 32

# 3. Tampilkan hasilnya
print(f"Suhu {celcius}°C sama dengan {fahrenheit}°F")
```

```python

#@title    🛒 Aplikasi Kasir Kantin RPL
nama_pembeli = "tolaan" #@param {type:"string"}
menu_makanan = "Mie Ayam Bakso" #@param ["Nasi Goreng Spesial", "Mie Ayam Bakso", "Soto Ayam"]
harga_satuan = 13000 #@param {type:"integer"}
jumlah_porsi = 5 #@param {type:"slider", min:1, max:10, step:1}
uang_dibayar = 100000 #@param {type:"integer"}

# LENGKAPI RUMUS DI BAWAH INI:
total_belanja = harga_satuan * jumlah_porsi
uang_kembalian = uang_dibayar - total_belanja

# Menampilkan Struk Belanja
print("======================================")
print("         STRUK KANTIN SMK RPL          ")
print("======================================")
print(f"Nama Pembeli   : {nama_pembeli}")
print(f"Menu Pesanan   : {menu_makanan}")
print(f"Harga Satuan   : Rp{harga_satuan}")
print(f"Jumlah Porsi   : {jumlah_porsi} porsi")
print("--------------------------------------")
print(f"TOTAL BELANJA : Rp{total_belanja}")
print(f"UANG TUNAI     : Rp{uang_dibayar}")
print(f"UANG KEMBALIAN : Rp{uang_kembalian}")
print("======================================")
```
